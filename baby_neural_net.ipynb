### Neural Network Weights and Biases

Here we see what happens when we change the weights and biases of a neural network. A more complex approach involves backpropagation

```python
import numpy as np
np.set_printoptions(precision=3)
```

```python
def perceptron_layer(x, weight, bias):
    # YOUR CODE GOES HERE
    # # x is an N x n matrix of N inputs, each with n features
    # # an m x n weight matrix, to be multiplied by x
    # # bias, 1-D array of m biases, to be added to the m outputs

    # Note to self indexing is (row, column)
    weightTrans = x@weight.T
    weightTrans = weightTrans.T # Transpose the matrix

    m = len(bias) # length of rows
    N = len(x) # length of columns

    app = np.zeros((m,N))
    for i in range(N):
        app[:,i] = weightTrans[:,i] + bias.T

    a = app.T

    return a

 # Example: N = 3, n = 2, m = 4
x = np.array([[1,2],[3,4],[5,6]])
weight = np.array([[-1.5, -3], [0.5, 1], [1, 1.5], [2, -2]])
bias = np.array([3, -2, .5, -1])
a = perceptron_layer(x, weight, bias)
result = np.array(np.array([[ -4.5, 0.5, 4.5, -3. ],[-13.5, 3.5, 9.5, -3. ],[-22.5, 6.5, 14.5, -3. ]]))
print("Your result", a, sep="\n")
print("Correct result:", result, sep="\n")
```

```python
def sigmoid(x):
    return 1./(1.+np.exp(-x))
def MLP(x, weights, biases):
    # YOUR CODE GOES HERE
    m = len(weights)
    mlpf = []

    layer0 = perceptron_layer(x, weights[0], biases[0])
    actlayer0 = sigmoid(layer0) # initial activation layer
    actlayer = actlayer0

    for j in range(1,m-1):
        layer = perceptron_layer(actlayer, weights[j], biases[j])
        actlayer = sigmoid(layer) # activate the layer

    # finalize the loop
    layerf = perceptron_layer(actlayer,weights[m-1],biases[m-1])
    mlpf = layerf

    return mlpf
```

```python
# Example
np.random.seed(0)
dims = [2,6,8,3,1]
weights = []
biases = []
for i,_ in enumerate(dims[:-1]):
 weights.append(np.random.standard_normal([dims[i+1],dims[i]]))
 biases.append(np.random.rand(dims[i+1]))
x = np.random.uniform(-10,10,size=[10,2])
result = np.array([[0.029],[0.267],[0.314],[0.027],[0.319],[0.297],[0.331],[0.343],[0.187],[0.335]])
y = MLP(x, weights, biases)
print(" Your result: ", y.T, ".T",sep="")
print("Correct result: ", result.T, ".T", sep="")
```
